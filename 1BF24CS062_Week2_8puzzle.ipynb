```python
import random


def print_puzzle(state):
    for i in range(0, 9, 3):
        print(state[i:i + 3])
    print()


def get_neighbors(state):
    neighbors = []

    zero = state.index(0)
    row, col = divmod(zero, 3)

    moves = [
        (-1, 0),
        (1, 0),
        (0, -1),
        (0, 1)
    ]

    for dr, dc in moves:
        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_zero = new_row * 3 + new_col

            new_state = state.copy()

            new_state[zero], new_state[new_zero] = (
                new_state[new_zero],
                new_state[zero]
            )

            neighbors.append(new_state)

    return neighbors


def random_initial_state(goal, moves=10):
    state = goal.copy()
    previous = None

    for _ in range(moves):
        neighbors = get_neighbors(state)

        if previous is not None:
            filtered = [n for n in neighbors if n != previous]

            if filtered:
                neighbors = filtered

        previous = state
        state = random.choice(neighbors)

    return state


def depth_limited_search(
        state,
        goal,
        depth,
        path,
        path_set):

    if state == goal:
        return path

    if depth == 0:
        return None

    state_tuple = tuple(state)
    path_set.add(state_tuple)

    for neighbor in get_neighbors(state):

        neighbor_tuple = tuple(neighbor)

        if neighbor_tuple not in path_set:

            result = depth_limited_search(
                neighbor,
                goal,
                depth - 1,
                path + [neighbor],
                path_set
            )

            if result is not None:
                return result

    path_set.remove(state_tuple)

    return None


def ids(initial, goal):

    depth = 0

    while True:

        path_set = set()

        solution = depth_limited_search(
            initial,
            goal,
            depth,
            [initial],
            path_set
        )

        if solution is not None:
            return solution

        depth += 1



goal = [
    1, 2, 3,
    4, 5, 6,
    7, 8, 0
]

initial = random_initial_state(goal, 10)

print("Initial State:")
print_puzzle(initial)

print("Goal State:")
print_puzzle(goal)

solution = ids(initial, goal)

print("IDS Solution:")

if solution:
    for i, state in enumerate(solution):
        print("Step", i)
        print_puzzle(state)
else:
    print("Solution not found.")
```

```python
import random
from collections import deque

GOAL = (1, 2, 3, 4, 5, 6, 7, 8, 0)
DIRECTIONS = ((-1, 0), (1, 0), (0, -1), (0, 1))


def print_puzzle(state):
    for i in range(0, 9, 3):
        print(" ".join(str(t) if t else "_" for t in state[i:i + 3]))
    print()


def get_neighbors(state):
    blank = state.index(0)
    row, col = divmod(blank, 3)
    for dr, dc in DIRECTIONS:
        r, c = row + dr, col + dc
        if 0 <= r < 3 and 0 <= c < 3:
            swap = r * 3 + c
            s = list(state)
            s[blank], s[swap] = s[swap], s[blank]
            yield tuple(s)


def random_initial_state(goal=GOAL, moves=10, seed=None):
    rng = random.Random(seed)
    state, previous = goal, None
    for _ in range(moves):
        options = [n for n in get_neighbors(state) if n != previous]
        previous, state = state, rng.choice(options)
    return state


def solve_bfs(initial, goal=GOAL):
    parents = {initial: None}
    queue = deque([initial])
    while queue:
        state = queue.popleft()
        if state == goal:
            path = []
            while state is not None:
                path.append(state)
                state = parents[state]
            return path[::-1]
        for n in get_neighbors(state):
            if n not in parents:
                parents[n] = state
                queue.append(n)
    return None


if __name__ == "__main__":
    initial = random_initial_state(moves=10)
    solution = solve_bfs(initial)
    if solution:
        print(f"Solved in {len(solution) - 1} moves\n")
        for state in solution:
            print_puzzle(state)
    else:
        print("No solution")
```
